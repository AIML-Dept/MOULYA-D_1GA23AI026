# **Implement a CNOT gate circuit with the control qubit in state |1> and verify the target qubit flips**.

```python
!pip install qiskit qiskit-aer

from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)

# Put control qubit q0 into |1>
qc.x(0)

# CNOT: q0 = control, q1 = target
qc.cx(0, 1)

# Measure both qubits
qc.measure([0, 1], [0, 1])

print(qc)

simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()
print(result.get_counts())
```

# Build a Bell state circuit and run 1024 shots, verifying the outcomes are only 00 and 11.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(2, 2)

qc.h(0)
qc.cx(0, 1)

qc.measure([0, 1], [0, 1])

print(qc)

simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()

print(counts)
```

# Implement a 3-qubit GHZ state and confirm perfect three-way correlation in measurement results.

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

qc = QuantumCircuit(3, 3)

qc.h(0)
qc.cx(0, 1)
qc.cx(1, 2)

qc.measure([0, 1, 2], [0, 1, 2])

print(qc)

simulator = AerSimulator()
result = simulator.run(qc, shots=1024).result()
counts = result.get_counts()

print(counts)
```

# Emulate a classical full-adder circuit using Toffoli and CNOT gates.

```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(5, 2)

# q0 = A
# q1 = B
# q2 = Cin
# q3 = Sum
# q4 = Carry

# Sum = A XOR B XOR Cin
qc.cx(0, 3)
qc.cx(1, 3)
qc.cx(2, 3)

# Carry = AB XOR Cin(A XOR B)
qc.ccx(0, 1, 4)
qc.cx(0, 3)
qc.ccx(3, 2, 4)
qc.cx(0, 3)

qc.measure(3, 0)
qc.measure(4, 1)

print(qc.draw())
```
