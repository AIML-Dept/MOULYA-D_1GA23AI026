```python
from itertools import product

def xor_bits(a, b):
    return ''.join(str(int(x) ^ int(y)) for x, y in zip(a, b))


def simon_oracle(x, secret):
    """
    Simon oracle satisfying:
        f(x) = f(x XOR secret)
    """

    y = xor_bits(x, secret)

    # Same representative for x and x XOR secret
    representative = min(x, y)

    return representative


secret = "110"

print("SIMON'S ORACLE")
print("Secret =", secret)
print("-" * 50)

print(f"{'x':<10}{'x XOR s':<12}{'f(x)':<10}{'f(x XOR s)':<15}{'Correct'}")

for bits in product("01", repeat=3):

    x = ''.join(bits)
    y = xor_bits(x, secret)

    fx = simon_oracle(x, secret)
    fy = simon_oracle(y, secret)

    correct = (fx == fy)

    print(
        f"{x:<10}"
        f"{y:<12}"
        f"{fx:<10}"
        f"{fy:<15}"
        f"{correct}"
    )

print("\nOracle correctness verified!")
```

```python
import numpy as np
from itertools import product


def dot_mod2(a, b):
    return sum(int(x) * int(y) for x, y in zip(a, b)) % 2


def gf2_rank(rows):

    if not rows:
        return 0

    A = np.array(
        [[int(bit) for bit in row] for row in rows],
        dtype=np.uint8
    )

    rows_count, cols = A.shape
    rank = 0

    for col in range(cols):

        pivot = None

        for row in range(rank, rows_count):
            if A[row, col] == 1:
                pivot = row
                break

        if pivot is not None:

            A[[rank, pivot]] = A[[pivot, rank]]

            for row in range(rows_count):

                if row != rank and A[row, col]:
                    A[row] ^= A[rank]

            rank += 1

    return rank


def generate_measurements(secret):

    n = len(secret)
    equations = []

    candidates = [
        ''.join(bits)
        for bits in product("01", repeat=n)
        if ''.join(bits) != "0"
    ]

    for y in candidates:

        if dot_mod2(y, secret) == 0:

            test = equations + [y]

            if gf2_rank(test) > gf2_rank(equations):
                equations.append(y)

        if len(equations) >= n - 1:
            break

    return equations


# 3-bit secret
secret = "110"

equations = generate_measurements(secret)

print("SIMON'S ALGORITHM — 3 BIT")
print("Secret =", secret)
print("-" * 50)

for i, y in enumerate(equations, 1):

    result = dot_mod2(y, secret)

    print(
        f"Equation {i}: "
        f"{y} . {secret} = {result} (mod 2)"
    )

print("\nIndependent equations =", gf2_rank(equations))

if gf2_rank(equations) >= len(secret) - 1:
    print("SUCCESS: n-1 independent equations obtained.")
```

```python
import numpy as np
from itertools import product


# --------------------------------------------------
# Binary inner product
# --------------------------------------------------

def dot_mod2(a, b):
    return sum(int(x) * int(y) for x, y in zip(a, b)) % 2


# --------------------------------------------------
# GF(2) Rank
# --------------------------------------------------

def gf2_rank(rows):

    if not rows:
        return 0

    A = np.array(
        [[int(bit) for bit in row] for row in rows],
        dtype=np.uint8
    )

    m, n = A.shape
    rank = 0

    for col in range(n):

        pivot = None

        for row in range(rank, m):

            if A[row, col] == 1:
                pivot = row
                break

        if pivot is not None:

            A[[rank, pivot]] = A[[pivot, rank]]

            for row in range(m):

                if row != rank and A[row, col]:
                    A[row] ^= A[rank]

            rank += 1

    return rank


# --------------------------------------------------
# Generate independent equations
# --------------------------------------------------

def generate_equations(secret):

    n = len(secret)
    equations = []

    for bits in product("01", repeat=n):

        y = ''.join(bits)

        if y == "0":
            continue

        # Simon measurement condition
        if dot_mod2(y, secret) == 0:

            candidate = equations + [y]

            if gf2_rank(candidate) > gf2_rank(equations):
                equations.append(y)

        if len(equations) == n - 1:
            break

    return equations


# --------------------------------------------------
# Find GF(2) nullspace
# --------------------------------------------------

def gf2_nullspace(equations, n):

    A = np.array(
        [[int(bit) for bit in row] for row in equations],
        dtype=np.uint8
    )

    m = len(equations)

    pivot_columns = []
    row = 0

    for col in range(n):

        pivot = None

        for r in range(row, m):

            if A[r, col]:
                pivot = r
                break

        if pivot is None:
            continue

        A[[row, pivot]] = A[[pivot, row]]

        for r in range(m):

            if r != row and A[r, col]:
                A[r] ^= A[row]

        pivot_columns.append(col)
        row += 1

    free_columns = [
        c for c in range(n)
        if c not in pivot_columns
    ]

    basis = []

    for free in free_columns:

        vector = np.zeros(n, dtype=np.uint8)

        vector[free] = 1

        for r in range(len(pivot_columns) - 1, -1, -1):

            pivot = pivot_columns[r]

            value = 0

            for c in free_columns:

                if A[r, c] and vector[c]:
                    value ^= 1

            vector[pivot] = value

        basis.append(
            ''.join(str(int(x)) for x in vector)
        )

    return basis


# ==================================================
# MAIN
# ==================================================

secret = "1011"

equations = generate_equations(secret)

print("SIMON'S ALGORITHM — 4 BIT")
print("Actual secret =", secret)
print("-" * 50)

print("\nMeasurement equations:")

for i, y in enumerate(equations, 1):

    print(
        f"Equation {i}: "
        f"{y} . s = 0 (mod 2)"
    )

print("\nRank =", gf2_rank(equations))

# Solve system
solutions = gf2_nullspace(
    equations,
    len(secret)
)

print("\nGF(2) Nullspace:")

for solution in solutions:
    print(solution)

if len(solutions) == 1:

    recovered = solutions[0]

    print("\nRecovered secret =", recovered)

    if recovered == secret:
        print("SUCCESS: Secret recovered correctly!")

else:

    print(
        "\nThe equations do not yet uniquely determine "
        "the non-zero secret."
    )
```

```python

import matplotlib.pyplot as plt


# Secret-string lengths
n_values = list(range(2, 21))


# --------------------------------------------------
# Oracle queries
# --------------------------------------------------

# Classical brute-force:
# approximately 2^(n-1)

classical_queries = [
    2 ** (n - 1)
    for n in n_values
]


# Simon's algorithm:
# approximately n oracle queries

simon_queries = [
    n
    for n in n_values
]


# --------------------------------------------------
# Calculate speedup
# --------------------------------------------------

speedup = [
    classical / simon
    for classical, simon
    in zip(classical_queries, simon_queries)
]


# --------------------------------------------------
# Print results
# --------------------------------------------------

print("SIMON'S ALGORITHM VS CLASSICAL SEARCH")
print("-" * 60)

print(
    f"{'n':<5}"
    f"{'Classical Queries':<20}"
    f"{'Simon Queries':<15}"
    f"{'Speedup':<15}"
)

print("-" * 60)

for n, classical, simon, speed in zip(
    n_values,
    classical_queries,
    simon_queries,
    speedup
):

    print(
        f"{n:<5}"
        f"{classical:<20}"
        f"{simon:<15}"
        f"{speed:.2f}x"
    )


# --------------------------------------------------
# Plot query complexity
# --------------------------------------------------

plt.figure(figsize=(9, 5))

plt.semilogy(
    n_values,
    classical_queries,
    marker="o",
    label="Classical brute force"
)

plt.semilogy(
    n_values,
    simon_queries,
    marker="s",
    label="Simon's algorithm"
)

plt.xlabel("Secret string length (n)")
plt.ylabel("Number of oracle queries")
plt.title("Simon’s Algorithm vs Classical Brute Force")

plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()


# --------------------------------------------------
# Plot speedup
# --------------------------------------------------

plt.figure(figsize=(9, 5))

plt.plot(
    n_values,
    speedup,
    marker="o"
)

plt.xlabel("Secret string length (n)")
plt.ylabel("Speedup")
plt.title("Simon’s Algorithm Theoretical Speedup")

plt.grid(True)

plt.tight_layout()
plt.show()
```
